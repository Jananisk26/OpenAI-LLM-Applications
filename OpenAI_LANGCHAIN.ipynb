```python
# Import libraries
import os     # Work with environment variables and operating system functions
from getpass import getpass   #Securely accept password or API key input
from langchain_openai import ChatOpenAI   #Import the ChatOpenAI class to connect LangChain with OpenAI models
```

```python
# Enter your OpenAI API key
os.environ["OPENAI_API_KEY"]= getpass("Enter your OpenAI API key: ")
```

```python
# Create the LLM
llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0.7
)
```

```python
# Start chatbot

print("\n AI Chatbot Started!")  #It displays that the chatbot has started.
# this prints, "🤖 AI Chatbot Started!".
# \n creates a new line

print("Type 'exit' to stop.\n")
# this prints, "Type 'exit' to stop".

while True:  # while executes a block of code repeatedly.
             # True means the condition is always true.
             # So the chatbot keeps running continuously
    question = input("You: ")

    #stop chatbot
    if question.lower()=="exit":   # gets information from the user.The entered question is stored in the variable question
                                   # This checks whether the user entered exit
                                   # if checks a condition.
                                   # .lower() converts the input into lowercase
                                   # == checks whether two values are equal.

        print("\nChatbot stopped") #If the user enters exit, this message is displayed
        break #stops the loop


    #Send question to LLM
    response = llm.invoke(question)  # llm represents your language model.
                                     # .invoke() sends the user's question to the language model.
                                     # The AI processes the question and generates an answer.
                                     # The answer is stored in response


    #Display response
    print("\nAI:", response.content)  # response contains the model's response object.
                                      # .content extracts the actual text answer.
    print()

```
